//Retorna un arreglo con los valores únicos contenidos en la columna 'Gender'

```python
df.pivot_table(values = "Amount", index = "Gender", aggfunc="sum")
df.pivot_table(values = "Amount", index = "Gender", aggfunc="mean")
```

//Crea una tabla dinámica que calcula la suma total del monto ('Amount') agrupado por género ('Gender')
//Crea una tabla dinámica que calcula el promedio del monto ('Amount') agrupado por género ('Gender')

```python
df.pivot_table(values = "Amount", index = "Menu", aggfunc="sum")
```

//crea una tabla dinámica que calcula la suma total del monto ('Amount') agrupado por cada tipo de menú ('Menu')

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], aggfunc="sum")
```

//Crea una tabla dinámica multinivel agrupando por género ('Gender') y luego por menú ('Menu') para sumar el monto

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = "City", aggfunc="sum")
```

//Genera una tabla dinámica bidimensional cruzando género y menú en las filas contra las ciudades ('City') en las columnas

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")
```
