```python
import pandas as pd
customers = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Customers.csv')
sales = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Sales.csv')
```

//Lee el archivo CSV de clientes desde la ruta especificada y lo guarda en el DataFrame 'customers'
//Lee el archivo CSV de ventas desde la ruta especificada y lo guarda en el DataFrame 'sales'

```python
customers
```

//Muestra el contenido completo del DataFrame de clientes

```python
sales
```

//Muestra el contenido completo del DataFrame de ventas

```python
cust_sales = pd.merge(customers, sales, how = 'inner', on = 'cust_id')
cust_sales
```

//Combina 'customers' y 'sales' en un 'inner join' usando la columna 'cust_id' como clave común
//Imprime/muestra el resultado del DataFrame resultante 'cust_sales'

```python
#inner join
cust_sales = pd.merge(customers, sales[['cust_id', 'product', 'qty', 'store_location']], how = 'inner', on = 'cust_id')
cust_sales
```

//Realiza un 'right join' manteniendo todos los registros de la tabla 'sales', tengan o no cliente coincidente

```python
pd.merge(customers, sales, how = "outer", on = 'cust_id')
```

//Realiza un 'outer join' (Full Outer Join) incluyendo todos los registros de ambas tablas aunque no coincidan

```python
products = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Products.csv')
products
```

//Lee el archivo CSV de productos desde la ruta especificada y lo almacena en el DataFrame 'products'

```python
pd.merge(customers, sales, how = 'inner', on = 'cust_id')
```

//Realiza nuevamente un 'inner join' simple entre clientes y ventas

```python
customers.merge(sales, on = 'cust_id').head(2)
```

//Usa la sintaxis orientada a objetos 'merge' para unir 'customers' y 'sales', mostrando solo los primeros 2 registros

```python
customers.merge(sales, on = 'cust_id').merge(products, on = 'prod_id').head(2)
```

//Encadena dos uniones: une 'customers' con 'sales' y luego el resultado con 'products' usando 'prod_id', mostrando 2 filas

```python
customers.merge(sales, on = 'cust_id').merge(products[['prod_id', 'price']], on = 'prod_id').head(2)
```
