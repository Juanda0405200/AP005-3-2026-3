```python
for i in range(1, 6):  # Inicia el bucle 'for' asignando a 'i' valores del 1 al 5.
    while i <= 4:  # Mantiene un bucle mientras 'i' sea menor o igual a 4.
        i += 1  # Incrementa el valor de 'i' en 1 en cada repetición.
        print(i)  # Muestra el valor actualizado de '
        i' en pantalla.
    break  # Cancela el bucle 'for' inmediatamente, impidiendo que pase a los siguientes números (2, 3, 4, 5).
```
