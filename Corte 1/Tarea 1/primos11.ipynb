```python
import time
inicio = time.time() #Guarda el tiempo inicial.

for i in range(1,31):
    conta = 0
    for n in range(1, i+1): #Divide i entre todos los números enteros desde $1$ hasta $i$
        residue = i%n #Obtiene el resto de la división.
        if residue == 0: #si no hay residuo, incrementa el contador de divisores.
            conta = conta + 1 #Si el número solo tiene 2 divisores, entra al bloque:
    if conta == 2:
        print(f'{i} es un primo') #Imprime el número primo.
        print("\n")

fin = time.time()
print("t = ", (fin - inicio)*1000)
```
