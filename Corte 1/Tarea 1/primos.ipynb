```python
import time
inicio = time.time()
# guarda el tiempo en cual empieza el codigo

for i in range(0,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n #Calcula el residuo de dividir i entre n
        if residue == 0: #Si la división es exacta (residuo 0), suma 1 al contador de divisores.
            conta = conta + 1

    if conta == 2:
        print(f'{i} es un primo')

fin = time.time() #Toma el tiempo al finalizar todo el proceso.
print("t = ", (fin - inicio)*1000) #Resta el tiempo inicial del final para obtener la duración en segundos y la muestra en ms
```
